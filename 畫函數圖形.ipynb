```python
%matplotlib inline

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

### 創意圖形一：對數螺線 (Logarithmic Spiral)

對數螺線是一種特殊的螺線，其半徑隨著角度的增加呈指數增長。在自然界中，許多現象都呈現出對數螺線的形狀，例如鸚鵡螺的殼、向日葵的種子排列以及星系的旋臂等。其在極座標系中的方程式通常表示為 $r = a \cdot e^{b\theta}$，其中 $a$ 和 $b$ 是常數，$\theta$ 是角度。

在這個範例中，我將使用 $a=0.1$ 和 $b=0.3$ 來繪製一個美麗的對數螺線。

```python
# 定義對數螺線的參數
a = 0.1
b = 0.3

# 生成角度 theta 的範圍
theta = np.linspace(0, 10 * np.pi, 1000) # 從 0 到 10π，生成 1000 個點

# 計算極徑 r
r = a * np.exp(b * theta)

# 將極座標轉換為笛卡爾座標
x_spiral = r * np.cos(theta)
y_spiral = r * np.sin(theta)

# 繪製對數螺線
plt.figure(figsize=(8, 8))
plt.plot(x_spiral, y_spiral, color='blue')
plt.title(r'Logarithmic Spiral ($r = 0.1 \cdot e^{0.3\theta}$)') # 使用原始字串消除警告
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.axhline(0, color='black',linewidth=0.5)
plt.axvline(0, color='black',linewidth=0.5)
plt.gca().set_aspect('equal', adjustable='box') # 保持x軸和y軸比例一致，使圖形不變形
plt.show()
```

### 創意圖形二：玫瑰曲線 (Rose Curve)

玫瑰曲線，又稱為萊洛多線 (Rhodonea curve)，是一種極座標下的數學曲線，其形狀像一朵花。它的極座標方程式為 $r = \cos(k\theta)$ 或 $r = \sin(k\theta)$。其中 $k$ 是一個決定花瓣數量的參數。

*   如果 $k$ 是整數：
    *   如果 $k$ 是奇數，花瓣的數量就是 $k$。
    *   如果 $k$ 是偶數，花瓣的數量就是 $2k$。
*   如果 $k$ 是分數 $p/q$：
    *   如果 $q$ 是奇數，花瓣的數量是 $p$。
    *   如果 $q$ 是偶數，花瓣的數量是 $2p$。

在這個範例中，我將使用 $r = \cos(4\theta)$ 來繪製一個有 8 片花瓣的玫瑰曲線 (因為 $k=4$ 是偶數)。

```python
# 定義玫瑰曲線的參數
k = 4 # 決定花瓣數量的參數，由於 k 是偶數，花瓣數量會是 2k = 8

# 生成角度 theta 的範圍
theta_rose = np.linspace(0, 2 * np.pi, 1000) # 從 0 到 2π，生成 1000 個點

# 計算極徑 r
r_rose = np.cos(k * theta_rose)

# 將極座標轉換為笛卡爾座標
x_rose = r_rose * np.cos(theta_rose)
y_rose = r_rose * np.sin(theta_rose)

# 繪製玫瑰曲線
plt.figure(figsize=(8, 8))
plt.plot(x_rose, y_rose, color='green')
plt.title(r'Rose Curve ($r = \cos(4\theta)$) - 8 Petals') # 使用原始字串消除警告
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.axhline(0, color='black',linewidth=0.5)
plt.axvline(0, color='black',linewidth=0.5)
plt.gca().set_aspect('equal', adjustable='box') # 保持x軸和y軸比例一致
plt.show()
```
